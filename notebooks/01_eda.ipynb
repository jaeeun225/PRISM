```python
import pandas as pd

df = pd.read_csv('data/raw/fragella_raw.csv', low_memory=False)
print(f'총 행수: {len(df)}')
print(f'컬럼 수: {len(df.columns)}')
df.head(3)
```

```python
# 결측률 확인
print('=== 컬럼별 결측률 ===')
for col in df.columns:
    null_n = df[col].isna().sum()
    null_pct = null_n / len(df) * 100
    print(f'{col}: {null_n}개 ({null_pct:.1f}%)')
```

```python
# 컬럼 타입 확인
print(df.dtypes)
```

```python
# 주요 컬럼 분포 확인
print('=== Year 분포 (상위 20개) ===')
print(df['Year'].value_counts().head(20))

print('\n=== Gender 분포 ===')
print(df['Gender'].value_counts())

print('\n=== Popularity 분포 ===')
print(df['Popularity'].value_counts())

print('\n=== Confidence 분포 ===')
print(df['Confidence'].value_counts())
```

```python
# Year 연대별 분포
df['Year_num'] = pd.to_numeric(df['Year'], errors='coerce')

bins   = [1800, 1900, 1950, 1970, 1980, 1990, 2000, 2005, 2010, 2015, 2020, 2027]
labels = ['~1900', '1900s', '1950s', '1970s', '1980s', '1990s', '2000~04', '2005~09', '2010~14', '2015~19', '2020~']

era = pd.cut(df['Year_num'], bins=bins, labels=labels)
print('=== 연대별 향수 수 ===')
print(era.value_counts().sort_index())
print(f'\nMin: {df["Year_num"].min():.0f}, Max: {df["Year_num"].max():.0f}')
print(f'Unknown: {df["Year_num"].isna().sum()}개 ({df["Year_num"].isna().sum()/len(df)*100:.1f}%)')

df.drop(columns=['Year_num'], inplace=True)
```

```python
# 중복 현황 확인
print(f'전체 행수: {len(df)}')
print(f'_id 기준 중복: {df.duplicated(subset=["_id"]).sum()}개')
print(f'Name+Brand 기준 중복: {df.duplicated(subset=["Name", "Brand"]).sum()}개')

print('\n=== 중복이 많은 브랜드 상위 10개 ===')
dup_by_brand = df[df.duplicated(subset=['_id'], keep=False)].groupby('Brand').size()
print(dup_by_brand.sort_values(ascending=False).head(10))
```

```python
# 중복 원인 분석 - Anna Sui 케이스
import json

with open('data/raw/collection_log.json', 'r', encoding='utf-8') as f:
    log = json.load(f)

anna = df[df['Brand'] == 'Anna Sui']
print(f'Anna Sui 전체 행수: {len(anna)}')
print(f'Anna Sui 고유 향수 수 (_id 기준): {anna["_id"].nunique()}')
print(f'collection_log 수집 기록: {log["brand_counts"].get("Anna Sui", 0)}개 (1회 호출)')

id_counts = anna['_id'].value_counts()
print(f'\n중복 횟수 — 최소: {id_counts.min()}회 / 최대: {id_counts.max()}회 / 평균: {id_counts.mean():.1f}회')

print('\n=== 중복 횟수별 향수 수 ===')
print('중복 횟수  향수 수')
for cnt, n in id_counts.value_counts().sort_index().items():
    print(f'{cnt:<10} {n}')
```

```python
# Anna Sui 그룹 분석
anna_indices = sorted(df[df['Brand'] == 'Anna Sui'].index.tolist())

chunks = []
chunk_start = anna_indices[0]
prev = anna_indices[0]

for idx in anna_indices[1:]:
    if idx - prev > 100:
        chunks.append((chunk_start, prev))
        chunk_start = idx
    prev = idx
chunks.append((chunk_start, prev))

print(f'Anna Sui 그룹 수: {len(chunks)}개 (collection_log 호출 기록: 1회)')
for i, (start, end) in enumerate(chunks):
    size = len([x for x in anna_indices if start <= x <= end])
    print(f'  그룹 {i+1}: 인덱스 {start}~{end}, {size}개')
```

**Anna Sui를 예시로 분석한 결과, 'collection_log 기준 1회 호출 / 50개 수집'임에도
실제 데이터에는 '14개 그룹 / 683행'이 존재한다.**

그룹 크기가 2~304개로 제각각이며, 직접 호출 시 매번 동일한 50개가 반환되는 것과 다르다.

이는 다른 브랜드 호출 시 Anna Sui 향수가 응답에 섞여 들어온 것으로 판단된다.

재수집해도 동일하게 발생하므로, 전처리 단계에서 `_id` 기준 중복 제거로 처리하기로 했다.
